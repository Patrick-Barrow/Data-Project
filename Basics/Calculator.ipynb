```python
def calculator (num1,operator,num2):
    if operator == '+':
        return(num1+num2)
    elif operator == '-':
        return(num1-num2)
    elif operator == '*':
        return(num1*num2)
    elif operator == '/':
        if num2 == 0:
            return('Cannot divide by 0')
        else:
            return(num1/num2)

def get_number(prompt):
    while True:
        num=input(prompt)
        try:
          return (float(num))
        except:
          print('Please enter a valid number!')

def get_operator():
    while True:
        operator = input('Please enter operator (+,-,*,/) here: ')
        if operator in ['+', '-', '*', '/']:
            return operator
        else:
            print('Please enter a valid operator!')

num1 = get_number('Please enter the first number here: ')
operator = get_operator()
num2 = get_number('Please enter the second number here: ')

result = calculator(num1,operator,num2)
if isinstance(result, float) and result.is_integer():
    print(int(result))
else:
    print(result)
```
