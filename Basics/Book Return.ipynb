```python
def get_book_title():
    while True:
        title = input('Enter book title: ')
        if title != '':
            return title
        print('Title cannot be empty!')


def get_days(prompt):
    while True:
        days = input(prompt)
        try:
            return int(days)
        except:
            print('Please enter a valid number!')


def calculate_fine(days_overdue):
    fine_per_day = 0.50
    return days_overdue * fine_per_day


def print_receipt(title, days_borrowed, fine):
    print(f'Book: {title}')
    print(f'Days borrowed: {days_borrowed}')
    print(f'Days overdue: {max(0, days_borrowed - 14)}')
    print(f'Fine: ${fine:.2f}')


# Main program
title = get_book_title()
days_borrowed = get_days('How many days was it borrowed? ')
days_overdue = max(0, days_borrowed - 14)
fine = calculate_fine(days_overdue)

print_receipt(title, days_borrowed, fine)
```
