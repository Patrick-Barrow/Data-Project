```python
while True:
    customer_name = input('Please enter your name here: ')
    if customer_name == 'done':
        break

    drink_count = 0
    total_price = 0
    orders = []
    print(drink_count, total_price, orders)
    print("Now taking drink order for", customer_name)
    while True:
        coffee_name = input("Please enter your drink order or type 'Finished' to end: ")
        if coffee_name.lower() == 'finished':
            break
        if coffee_name.lower() == 'latte':
            drink_count +=1
            total_price +=3.50
            orders.append('latte')
        elif coffee_name.lower() == 'americano':
            drink_count +=1
            total_price +=3.00
            orders.append('americano')
        elif coffee_name.lower() == 'espresso':
            drink_count +=1
            total_price +=2.50
            orders.append('espresso')
        else:
            print('Please enter a valid cofee drink from our menu.')

    print(f'{customer_name}\'s order: {', '.join(item.capitalize() for item in orders)}')
    print(f'You ordered {drink_count} coffees.')
    print(f'Your total is {total_price:.2f}!')
```
